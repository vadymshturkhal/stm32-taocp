$$ \large \text{Four magic integers:} $$

$
\large m \text{, the modulus; }   \hspace {2cm} 0 < m \\
\large a \text{, the multiplier;} \hspace {2cm} 0 \le m \\
\large c \text{, the increment;}  \hspace {2cm} 0 \le c < m \\
\large X_{0} \text{, the starting value;} \hspace {1cm} 0 \le X_{0} < m 
$

$ \large \text{The desired sequence of random numbers } \langle X_{n} \rangle \text{ is then obtained by setting}$

$$ \large X_{n + 1} = (aX_{n} + c) \text{ mod } m, \hspace {1cm} n \ge 0 \tag{1}$$

$ \text{Let's define b:}$
$$ \large b = a - 1 \large $$

$ \text{And for practical purpose assume:}$
$$ \large a \ge 2, \qquad b \ge 1 $$

$ \large \text{The goal is to prove a generalization of Eq.(1), which is:} $

$$ \large X_{n+k} = \bigl( a^kX_n + \frac{(a^k - 1)c}{b} \bigr) \mod m, \qquad k \ge 0,\ n \ge 0 \tag{2}$$

$ \text{And it expresses the } \textit{(n+k)} \text{th term directly in terms of the } \textit{n} \text{th term} $

$ \large \text{Eq. (1) tells that } X_{n+1} \text{ is } \textit{congruent} \text{ to } (aX_{n} + c) \text{, and it's in } (0,m] \text{ range:}$

$$ \large X_{n+1}\equiv (aX_{n} + c) \pmod m, \qquad 0 \le X_{n+1} < m$$

$ \text{and according to the fact: if } x\equiv y \text{ then } x + c \equiv y + c \text{, we can iterate without carrying a mod at every step}$

$ \large \text{k=3 example:} $

$$ \large X_{n+1}\equiv (aX_{n} + c)$$

$$ \large X_{n+2}\equiv a(aX_{n} + c) + c => X_{n+2}\equiv a^2X_n + ac + c$$

$$ \large X_{n+3}\equiv a(a^2X_n + ac + c) + c => X_{n+3}\equiv a^3X_n + a^2c + ac + c$$

$ \large \text{The pattern is: } $
$$\large X_{n+k} \equiv a^kX_n + c(1 + a^1 + a^2 + \dots + a^{k-1}) \qquad \pmod m \tag 3 $$

$ \large \text{Consider } (1 + a^1 + a^2 + \dots + a^{k-1}) \text{ geometric series:} \\ $
$$ \large 1 + a^1 + a^2 + \dots + a^{k-1} = \sum\limits_{k=0}^{n-1} 1 \cdot a^k = \frac{1 - a^k}{1 - a}$$

$ \text {Multiply by -1 for using b:} $

$$ \large \frac{1 - a^k}{1 - a} = \frac{-(1 - a^k)}{-(1 - a)} = \frac{a^k - 1}{a - 1}$$

$\text{Use b:} $

$$ \large \frac{a^k - 1}{a - 1} = \frac{a^k - 1}{b} \tag 4 $$

$ \text{Replace geometric series in Eq.(3) with retrieved Eq.(4)} $

$$\large X_{n+k} \equiv a^kX_n + \frac{(a^k - 1)c}{b} \qquad \pmod m $$

$ \text{Thus we proved Eq.(2)} $
